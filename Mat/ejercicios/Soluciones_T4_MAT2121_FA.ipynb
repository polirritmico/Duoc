```python
import numpy as np
```

#### Respuesta 3.

```python
# Cols: Medallas de oro, plata, bronce
matrixF_suma_medallas = np.array([
    [42, 43, 31],  # USA
    [40, 35, 20],  # China
    [30, 15, 16],  # Francia
])

matrixG_puntaje_medalla = np.array([
    [3],  # Oro
    [2],  # Plata
    [1],  # Bronce
])
```

```python
matrixH_puntajes = np.dot(matrixF_suma_medallas, matrixG_puntaje_medalla)

print("Matriz H:\n", matrixH_puntajes)
```

La diferencia $h_{11}-h_{21}$ representa la _diferencia de puntaje_ entre **USA** y **China**.

#### Respuesta 4.

```python
A = np.array([
    [1, 1, 1],
    [3, 2, 1],
    [53, 40, 20],
])
B = np.array([
    [10],
    [17],
    [326],
])

A_inv = np.linalg.inv(A)
X = np.dot(A_inv, B)

print(X)
```

```python
# Revisamos:
assert 326 == (53 * 2) + (40 * 3) + (20 * 5)
```

Resolviendo obtenemos:

- $x=2$
- $y=3$
- $z=5$

Donde $x$ es la cantidad de medallas de oro, $y$ es la cantidad de medallas de plata y $z$ la cantidad de medallas de bronce.
