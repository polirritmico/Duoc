### Respuesta 1


¿Cuál es el valor de $t$ a las 07:30 de la mañana? ¿Cuál sería la cantidad de agua acumulada en Santiago a las 07:30 de la mañana?

```python
def tiempo_a_hora(hora: int, minutos: int) -> float:
    minutos = (hora * 60) + minutos
    horas = minutos / 60
    return horas

def precipitacion_santiago(tiempo: float) -> float:
    resultado = (1.1 * tiempo ** 2) + (15.8 * tiempo) + 32
    return resultado
```

```python
tiempo_final = tiempo_a_hora(hora=1, minutos=30)
print(f"El valor de t a las 7:30 hrs. es: {tiempo_final} hrs.")

mm_caidos_santiago = precipitacion_santiago(tiempo_final)
print(f"La cantidad de agua acumulada en Santiago a las 07:30 hrs. es de {mm_caidos_santiago:.3f} mm.")
```

```python
def precipitacion_valparaiso(tiempo):
    resultado: float = (0.7 * tiempo ** 2) + (13.2 * tiempo) + 39
    return resultado

def precipitacion_santiago(tiempo):
    resultado: float = (1.1 * tiempo ** 2) + (15.8 * tiempo) + 32
    return resultado

def precipitacion_concepcion(tiempo):
    resultado: float = (1.3 * tiempo ** 2) + (18.9 * tiempo) + 28
    return resultado

def tiempo_hasta_mm_emergencia(mm_objetivo: int, funcion_precipitacion, step: float = 0.01) -> float:
    mm_actual = 0
    tiempo = 0
    while mm_actual < mm_objetivo:
        tiempo += step
        mm_actual = funcion_precipitacion(tiempo)
    return tiempo
```

```python
mm_emergencia: int = 100
precipitaciones = {
    "Santiago": tiempo_hasta_mm_emergencia(mm_emergencia, precipitacion_santiago),
    "Valparaíso": tiempo_hasta_mm_emergencia(mm_emergencia, precipitacion_valparaiso),
    "Concepción": tiempo_hasta_mm_emergencia(mm_emergencia, precipitacion_concepcion),
}

primera_ciudad: str = min(precipitaciones)
primera_ciudad_tiempo: float = precipitaciones[primera_ciudad]

msg = "La primera ciudad en requerir atención de emergencia (más de {} mm caídos) será {} en {:.2f} hrs."
print(msg.format(mm_emergencia, primera_ciudad, primera_ciudad_tiempo))
```

### Respuesta 4

Entre las 3 y 5 de la tarde, ¿en qué ciudad cayó más agua?

```python
def precipitacion_valparaiso(tiempo):
    resultado: float = (0.7 * tiempo ** 2) + (13.2 * tiempo) + 39
    return resultado

def precipitacion_santiago(tiempo):
    resultado: float = (1.1 * tiempo ** 2) + (15.8 * tiempo) + 32
    return resultado

def precipitacion_concepcion(tiempo):
    resultado: float = (1.3 * tiempo ** 2) + (18.9 * tiempo) + 28
    return resultado

def mm_por_rango_tiempo(
        funcion_mm_caidos, tiempo_inicial: int, tiempo_final: int
    ) -> float:
    mm_inicial = funcion_mm_caidos(tiempo_inicial)
    mm_final = funcion_mm_caidos(tiempo_final)
    return mm_final - mm_inicial
```

```python
inicio_medicion = 6
tiempo_inicial = 15 - inicio_medicion
tiempo_final = 17 - inicio_medicion

precipitaciones: dict[str, float] = {
    "Santiago": mm_por_rango_tiempo(precipitacion_santiago, tiempo_inicial, tiempo_final),
    "Valparaíso": mm_por_rango_tiempo(precipitacion_valparaiso, tiempo_inicial, tiempo_final),
    "Concepción": mm_por_rango_tiempo(precipitacion_concepcion, tiempo_inicial, tiempo_final),
}

for ciudad, mm_caidos in precipitaciones.items():
    print(f"Milímetros caídos en {ciudad}: {mm_caidos:.2f}")

print(f"\nEntre las 15 y 17 horas cayó más agua en {min(precipitaciones)}.")
```
