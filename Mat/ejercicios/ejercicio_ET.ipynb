#### 13.

**Sucesión geométrica**

$a_4 = 24$ \
$a_6 = 96$

$a_9 = x$

$a_m = a_n \cdot r^{(m-n)}$

$96 = 24 \cdot r^{6-4}$ \
$96/24 = r^2$ \
$4 = r^2$

> $r = 2$

---

$a_4 = a_1 \cdot 2^{4-1}$ \
$24 = a_1 \cdot 2^3$

> $a_1 = 3$

```python
def p13(n):
    return 3 * 2 ** (n - 1)

assert p13(4) == 24
assert p13(6) == 96
```

```python
print(p13(9))
```
